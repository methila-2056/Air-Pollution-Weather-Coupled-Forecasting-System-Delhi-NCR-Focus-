# AeroCast-NCR — Data Exploration

Explore the engineered featured dataset for Delhi NCR. The 120 MB `featured_dataset.csv` is committed in two parts; we load **part 1** here (header + first half).

> Tip: byte-for-byte reassembly instructions are in `data/processed/SPLIT_NOTE.txt`.


```python
import os
import pandas as pd

CSV = os.path.join("data", "processed", "featured_dataset_part1.csv")
df = pd.read_csv(CSV)
df.head()
```

```python
print(f"shape: {df.shape}")
print(f"stations: {df['station_name'].nunique()}")
print(f"date span: {df['timestamp'].min()} .. {df['timestamp'].max()}")
df[['timestamp', 'station_name', 'pm25', 'pm10', 'o3', 'no2', 'so2', 'co', 'aqi']].describe()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(12, 4))

df['aqi_category'] = pd.cut(
    df['aqi'],
    bins=[0, 50, 100, 200, 300, 400, 501],
    labels=['Good', 'Satisfactory', 'Moderate', 'Poor', 'Very Poor', 'Severe'],
    include_lowest=True,
)
df['aqi_category'].value_counts().sort_index().plot(kind='bar', ax=ax[0], title='AQI category counts')

g = df.groupby('station_name')['pm25'].mean().sort_values(ascending=False)
g.plot(kind='bar', ax=ax[1], title='Mean PM2.5 by station')
plt.tight_layout()
plt.show()
```

```python
null_frac = df.isna().mean().sort_values(ascending=False).head(8)
print("Top 8 columns by missing-value fraction (first half):")
print(null_frac.map(lambda x: f"{x:.2%}"))
```
